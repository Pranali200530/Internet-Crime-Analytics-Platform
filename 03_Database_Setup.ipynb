SQlit database setup


```python
import pandas as pd
import sqlite3
import os
```

```python
df = pd.read_csv("data/internet_crime.csv")

df.head()
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nColumns:")
print(df.columns.tolist())
```

```python
os.makedirs("database", exist_ok=True)
```

```python
database_path = "database/crime.db"

connection = sqlite3.connect(database_path)

print("Database connected successfully.")
```

```python
df.to_sql(
    "crime_records",
    connection,
    if_exists="replace",
    index=False
)

print("crime_records table created successfully.")
```

```python
query = "SELECT * FROM crime_records LIMIT 5"

database_data = pd.read_sql_query(
    query,
    connection
)

database_data
```

```python
count_query = """
SELECT COUNT(*) AS total_records
FROM crime_records
"""

count_result = pd.read_sql_query(
    count_query,
    connection
)

print(
    "Total records in database:",
    count_result.iloc[0]["total_records"]
)
```

```python
connection.close()

print("Database connection closed.")
```
