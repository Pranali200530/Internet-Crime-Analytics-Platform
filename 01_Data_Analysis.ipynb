```python
import plotly.express as px
```

```python
attack_counts = df["Attack Type"].value_counts().reset_index()

attack_counts.columns = ["Attack Type", "Number of Incidents"]

attack_counts
```

```python
fig = px.bar(
    attack_counts,
    x="Attack Type",
    y="Number of Incidents",
    title="Number of Cyber Incidents by Attack Type",
    text="Number of Incidents"
)

fig.show()
```

```python
year_counts = df["Year"].value_counts().sort_index().reset_index()

year_counts.columns = ["Year", "Number of Incidents"]

year_counts
```

```python
fig = px.line(
    year_counts,
    x="Year",
    y="Number of Incidents",
    markers=True,
    title="Cyber Incidents Over the Years"
)

fig.show()
```

```python
financial_by_attack = (
    df.groupby("Attack Type")["Financial Loss (in Million $)"]
    .sum()
    .reset_index()
)

financial_by_attack

fig = px.bar(
    financial_by_attack,
    x="Attack Type",
    y="Financial Loss (in Million $)",
    title="Financial Loss by Attack Type",
    text="Financial Loss (in Million $)"
)

fig.show()
```

```python
country_counts = df["Country"].value_counts().reset_index()

country_counts.columns = ["Country", "Number of Incidents"]

country_counts
```

```python
fig = px.bar(
    country_counts,
    x="Country",
    y="Number of Incidents",
    title="Cyber Incidents by Country",
    text="Number of Incidents"
)

fig.show()
```

```python
industry_counts = df["Target Industry"].value_counts().reset_index()

industry_counts.columns = ["Target Industry", "Number of Incidents"]

industry_counts
```

```python
fig = px.bar(
    industry_counts,
    x="Target Industry",
    y="Number of Incidents",
    title="Cyber Incidents by Target Industry",
    text="Number of Incidents"
)

fig.show()
```

```python
users_by_attack = (
    df.groupby("Attack Type")["Number of Affected Users"]
    .sum()
    .reset_index()
)

users_by_attack
```

```python
fig = px.bar(
    users_by_attack,
    x="Attack Type",
    y="Number of Affected Users",
    title="Affected Users by Attack Type",
    text="Number of Affected Users"
)

fig.show()
```

```python
source_counts = df["Attack Source"].value_counts().reset_index()

source_counts.columns = ["Attack Source", "Number of Incidents"]

source_counts
```

```python
fig = px.bar(
    source_counts,
    x="Attack Source",
    y="Number of Incidents",
    title="Cyber Incidents by Attack Source",
    text="Number of Incidents"
)

fig.show()
```

```python
vulnerability_counts = (
    df["Security Vulnerability Type"]
    .value_counts()
    .reset_index()
)

vulnerability_counts.columns = [
    "Security Vulnerability Type",
    "Number of Incidents"
]

vulnerability_counts
```

```python
fig = px.bar(
    vulnerability_counts,
    x="Security Vulnerability Type",
    y="Number of Incidents",
    title="Cyber Incidents by Security Vulnerability",
    text="Number of Incidents"
)

fig.show()
```

```python
defense_counts = (
    df["Defense Mechanism Used"]
    .value_counts()
    .reset_index()
)

defense_counts.columns = [
    "Defense Mechanism Used",
    "Number of Incidents"
]

defense_counts
```

```python
fig = px.bar(
    defense_counts,
    x="Defense Mechanism Used",
    y="Number of Incidents",
    title="Cyber Incidents by Defense Mechanism",
    text="Number of Incidents"
)

fig.show()
```

```python
resolution_by_attack = (
    df.groupby("Attack Type")["Incident Resolution Time (in Hours)"]
    .mean()
    .reset_index()
)

resolution_by_attack
```

```python
fig = px.bar(
    resolution_by_attack,
    x="Attack Type",
    y="Incident Resolution Time (in Hours)",
    title="Average Incident Resolution Time by Attack Type",
    text="Incident Resolution Time (in Hours)"
)

fig.show()
```
